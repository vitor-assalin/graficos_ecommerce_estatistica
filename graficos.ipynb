```python
import matplotlib.pyplot as plt
import pandas as pd
from pandas import read_csv

df = pd.read_csv("ecommerce_estatistica.csv")
plt.figure(figsize=(10,6))
plt.hist(df["Qtd_Vendidos_Cod"], bins=50, color="blue")
plt.title("Histograma - Quantidade de Vendidos")
plt.xlabel("Qtd_Vendidos")
plt.ylabel("Frequência")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv("ecommerce_estatistica.csv")
sns.jointplot(x="Preço", y="Qtd_Vendidos_Cod", data=df, kind="scatter")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv("ecommerce_estatistica.csv")

sns.heatmap(df[[ "Nota",
    "N_Avaliações",
    "Desconto",
    "Preço",
    "Qtd_Vendidos_Cod"]].corr(), annot=True)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("ecommerce_estatistica.csv")

plt.figure(figsize=(10,6))
df["Marca"].value_counts().head(10).plot(kind="bar", color="blue")
plt.title("Gráfico de Barra - Top 10 Marcas com Mais Produtos")
plt.xlabel("Marca")
plt.ylabel("Quantidade")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("ecommerce_estatistica.csv")
plt.figure(figsize=(10,6))
df["Gênero"].value_counts().head(3).plot.pie(autopct="%.1f%%", startangle=90)
plt.title("Gráfico de Pizza - Gênero")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv("ecommerce_estatistica.csv")
plt.figure(figsize=(10,6))
sns.kdeplot(df["Preço"], fill=True, color="blue")
plt.title("Gráfico de Densidade - Preço")
plt.xlabel("Preço")
plt.ylabel("Densidade")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv("ecommerce_estatistica.csv")
plt.figure(figsize=(10,6))
sns.regplot(x="Preço", y="Qtd_Vendidos_Cod", data=df, color="Red", scatter_kws={"alpha": 0.5, "color": "blue"})
plt.title("Gráfico de Regressão - Preço x Qtd_Vendidos")
plt.xlabel("Preço")
plt.ylabel("Qtd_Vendidos")
plt.show()
```
